```python
# Admissions Analysis Mini Project

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------
# Step 1: Create Sample Dataset
# ---------------------------------
data = {
    "Student_ID": range(1, 51),
    "Gender": np.random.choice(["Male", "Female"], 50),
    "Department": np.random.choice(["Engineering", "Science", "Arts", "Management"], 50),
    "Entrance_Score": np.random.randint(50, 100, 50),
    "Interview_Score": np.random.randint(40, 100, 50),
    "Admission_Status": np.random.choice(["Admitted", "Rejected"], 50, p=[0.6, 0.4])
}

df = pd.DataFrame(data)

print("===== FIRST 5 RECORDS =====")
print(df.head())

# ---------------------------------
# Step 2: Basic Analysis
# ---------------------------------
print("\n===== OVERALL STATISTICS =====")
print("Total Applicants:", len(df))
print("Total Admitted:", (df["Admission_Status"] == "Admitted").sum())
print("Total Rejected:", (df["Admission_Status"] == "Rejected").sum())

acceptance_rate = (df["Admission_Status"] == "Admitted").mean() * 100
print("Acceptance Rate: {:.2f}%".format(acceptance_rate))

# ---------------------------------
# Step 3: Department-wise Analysis
# ---------------------------------
dept_analysis = df.groupby("Department")["Admission_Status"].value_counts().unstack()
print("\n===== DEPARTMENT-WISE ANALYSIS =====")
print(dept_analysis)

# ---------------------------------
# Step 4: Gender-wise Analysis
# ---------------------------------
gender_analysis = df.groupby("Gender")["Admission_Status"].value_counts().unstack()
print("\n===== GENDER-WISE ANALYSIS =====")
print(gender_analysis)

# ---------------------------------
# Step 5: Visualization
# ---------------------------------

plt.figure(figsize=(15,4))

# Bar Chart - Department Applications
plt.subplot(1,3,1)
df["Department"].value_counts().plot(kind='bar', color='skyblue')
plt.title("Applications by Department")

# Pie Chart - Admission Status
plt.subplot(1,3,2)
df["Admission_Status"].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Admission Distribution")

# Histogram - Entrance Scores
plt.subplot(1,3,3)
plt.hist(df["Entrance_Score"], bins=10, color='lightgreen', edgecolor='black')
plt.title("Entrance Score Distribution")

plt.tight_layout()
plt.show()

print("\n===== ADMISSIONS ANALYSIS COMPLETE =====")
```
